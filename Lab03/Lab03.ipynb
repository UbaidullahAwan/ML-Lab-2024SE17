```
import numpy as np
import pandas as pd
pd.set_option('display.width', 140)

clean = pd.DataFrame({
    "Roll": ["SE-01", "SE-02", "SE-03", "SE-04", "SE-05", "SE-06", "SE-07", "SE-08"],
    "Name": ["Ali", "Ayesha", "Bilal", "Fatima", "Hamza", "Sana", "Usman", "Zainab"],
    "City": ["Muzaffarabad", "Mirpur", "Bagh", "Muzaffarabad", "Rawalakot", "Mirpur", "Kotli", "Muzaffarabad"],
    "Maths": [78, 91, 45, 85, 62, 73, 38, 88],
    "Physics": [65, 88, 52, 79, 48, 81, 41, 92],
    "Attendance": [92, 97, 70, 88, 75, 90, 60, 95]})

messy = pd.DataFrame({
    "Roll": ["SE-01", "SE-02", "SE-03", "SE-04", "SE-05", "SE-06", "SE-07", "SE-04", "SE-08"],
    "Name": ["Ali", "Ayesha", "Bilal", "Fatima", "Hamza", "Sana", "Usman", "Fatima", "Zainab"],
    "City": [" Muzaffarabad", "Mirpur", "Bagh", "MUZAFFARABAD", "Rawalakot", "Mirpur", "Kotli", "MUZAFFARABAD", "Muzafarabad"],
    "Maths": [78, 91, np.nan, 85, 62, 73, 380, 85, 88],
    "Physics": [65, 88, 52, 79, 48, 105, 41, 79, 92],
    "Attendance": ["92%", "97", "70", "88", "75%", "90", "60", "88", "95"]})
print(clean)
```

```
df = messy.copy()

df = df.drop_duplicates(subset='Roll', keep='first')

df['Roll'] = df['Roll'].str.strip().str.upper()
df['Name'] = df['Name'].str.strip().str.title()
df['City'] = df['City'].str.strip().str.title()

df['City'] = df['City'].replace({
    'Muzafarabad': 'Muzaffarabad'
})

df['Attendance'] = (
    df['Attendance']
    .astype(str)
    .str.replace('%', '', regex=False)
    .str.strip()
)

df['Attendance'] = pd.to_numeric(df['Attendance'], errors='coerce')

df['Maths'] = pd.to_numeric(df['Maths'], errors='coerce')
df['Physics'] = pd.to_numeric(df['Physics'], errors='coerce')

df.loc[~df['Maths'].between(0, 100), 'Maths'] = np.nan
df.loc[~df['Physics'].between(0, 100), 'Physics'] = np.nan

df.loc[~df['Attendance'].between(0, 100), 'Attendance'] = np.nan

df['Maths'] = df['Maths'].fillna(df['Maths'].median())
df['Physics'] = df['Physics'].fillna(df['Physics'].median())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].median())

print("Cleaned Dataset:")
display(df)

print("Missing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)

print("\nFinal Shape:")
print(df.shape)
```

## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```
print("(a) Missing values before cleaning:")
print(df.isnull().sum())
```

```
original_rows = len(df)
print(original_rows)
```

```
df = df.dropna(subset=['Brand', 'Model'])
removed_rows = original_rows -len(df)
print("\n Rows removed due to missing Brand or Modal:", removed_rows)

df['Registered City'] = df['Registered City'].fillna('Unknown')
df['Fuel'] = df['Fuel'].fillna('Unknown')
df['Transaction Type'] = df['Transaction Type'].fillna('Unknown')
```

```
print("(b) Task 3")
df['Registered City'] = df['Registered City'].str.strip().str.title()

df['Price'] = df['Price'].astype(str).str.replace('Rs', '', regex=False)
df['Price'] = df['Price'].str.replace(',', '', regex=False)
df['Price'] = df['Price'].str.strip()
```

```
print("(c) Task 3")
df['Price'] = df['Price'].astype('int64')
df['Year'] = df['Year'].astype('Int64')

print("\nData Types After Conversion:")
print(df.dtypes)
```

```
print("(d) Task 3")
cities_before = df['Registered City'].nunique()

df['Registered City'] = df['Registered City'].replace({
    'Sukkar': 'Sukkur'
})

df['Model'] = df['Model'].replace({
    'Corrolla Altis': 'Corolla Altis'
})

cities_after = df['Registered City'].nunique()

print("\nDifferent Cities Before Correction:", cities_before)
print("Different Cities After Correction:", cities_after)
```

```
print("(e) Task 3")
duplicates = df.duplicated().sum()

print("\nExact Duplicate Ads:", duplicates)

df = df.drop_duplicates()

print("Rows After Removing Duplicates:", len(df))
```

```
print("(f) Task 3")
invalid_kms = (
    (df['KMs Driven'] < 100) |
    (df['KMs Driven'] > 1000000) |
    (df['KMs Driven'].isin([12345, 123456]))
)

df.loc[invalid_kms, 'KMs Driven'] = np.nan


df = df[
    ~df['Price'].isin([12345678, 87654321])
]

df = df[
    ~(
        df['Brand'].isin(['Suzuki', 'Daihatsu']) &
        (df['Price'] > 5000000)
    )
]
```

```
print("\nFinal Dataset Shape:")
print(df.shape)

print("\nFinal Missing Values:")
print(df.isnull().sum())
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```
print("(a) Create Price_lakh column")
df['Price_lakh'] = df['Price'] / 100000
```

```
print("(b) Create Age column")
df['Age'] = 2020 - df['Year']
```

```
print("(c) Find the 5 most common brands")
print("Top 5 Most Common Brands:")
display(df['Brand'].value_counts().head(5))
```

```
print("(d) Find the 5 cities with the most ads")
print("Top 5 Cities:")
display(df['Registered City'].value_counts().head(5))
```

```
print("(e) Median Price_lakh for each fuel type, highest first")
print("Median Price by Fuel Type:")
fuel_prices = df.groupby('Fuel')['Price_lakh'].median().sort_values(ascending=False)
display(fuel_prices)
```

```
print("(f) Median Price_lakh of Corolla GLI for each year from 2010 to 2018")
print("Corolla GLI Median Price by Year:")

corolla = df[
    df['Model'].str.contains('Corolla GLI', case=False, na=False)
    & df['Year'].between(2010, 2018)
]

corolla_yearly = corolla.groupby('Year')['Price_lakh'].median().sort_index()
display(corolla_yearly)

# Calculate average yearly value decrease
if 2010 in corolla_yearly.index and 2018 in corolla_yearly.index:
    yearly_loss = (corolla_yearly.loc[2010] - corolla_yearly.loc[2018]) / 8
    print("Average yearly value decrease:", round(yearly_loss, 2), "lakh rupees")
else:
    print("2010 or 2018 data is missing, so yearly decrease cannot be calculated.")
```

```
print("(g) Show the 5 most expensive ads")
print("5 Most Expensive Ads:")
display(
    df.nlargest(5, 'Price')[
        ['Brand', 'Model', 'Year', 'Price', 'Registered City']
    ]
)
```

```
print("(h) Compare Cash vs Installment/Leasing")
print("Transaction Type Analysis:")

transaction_analysis = df.groupby('Transaction Type').agg(
    Number_of_Ads=('Price', 'count'),
    Median_Price=('Price_lakh', 'median')
).sort_values(by='Number_of_Ads', ascending=False)

display(transaction_analysis)
```
