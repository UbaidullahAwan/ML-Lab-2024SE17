8- Function: return True if the word "dog" is in a string (case-insensitive).

```python
def find_dog(string):
    return "dog" in string.lower()

print(find_dog("I have a DOG"))
print(find_dog("I have a cat"))
```

9- Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(string):
    return string.lower().count("dog")

print(count_dog("Dog is a dog and DOG is cute"))
```

10- caught_speeding(speed, is_birthday): ≤ 60 "No Ticket", 61–80 "Small Ticket", ≥ 81 "Big Ticket". On a birthday, allow 5 mph more.

```python
def caught_speeding(speed, is_birthday):
    if is_birthday:
        speed -= 5

    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"


print(caught_speeding(60, False))  
print(caught_speeding(65, True))   
print(caught_speeding(85, True))   
print(caught_speeding(90, True))   
```
